```python
# 1
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11)
y = x ** 2

plt.plot(x, y)

plt.title("Line Plot of y = x²")
plt.xlabel("X values")
plt.ylabel("Y values")

plt.show()
```

```python
#2

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2025 = [12000, 13500, 14200, 15000, 14800, 16200, 17500, 18000, 17200, 19000, 21000, 24000]

plt.figure(figsize=(10, 6))
plt.plot(months, sales_2025, marker='o', linestyle='-', color='b', label='Sales 2025')

plt.title("Monthly Sales in 2025")
plt.xlabel("Months")
plt.ylabel("Sales (USD)")
plt.grid(True, linestyle='--', alpha=0.6)


plt.show()
```

```python
#11
branch_data = {
    'branch': ['CSE', 'ECE', 'ME', 'CE'],
    'avg_marks': [85, 78, 72, 80]
}
df_branch = pd.DataFrame(branch_data)

min_value = df_branch['avg_marks'].min()

colors = ['red' if value == min_value else 'blue'
          for value in df_branch['avg_marks']]

plt.bar(df_branch['branch'], df_branch['avg_marks'], color=colors)

plt.title("Average Marks by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Marks")

plt.show()
```

```python
#12

student_data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah'],
    'marks': [82, 95, 71, 88, 97, 65, 90, 94]
}
df_students = pd.DataFrame(student_data)


top5 = df_students.sort_values(by='marks').tail(5)

plt.barh(top5['name'], top5['marks'], color='skyblue')

plt.title("Top 5 Students by Marks")
plt.xlabel("Marks")
plt.ylabel("Student Names")

plt.show()
```
