```python
#1
from datetime import datetime

# Create the datetime object representing 15 February 2026 at 10:30 AM
dt = datetime(2026, 2, 15, 10, 30)

# Display the components separately
print(f"Year: {dt.year}")
print(f"Month: {dt.month}")
print(f"Day: {dt.day}")
print(f"Hour: {dt.hour}")
print(f"Minute: {dt.minute}")
```

```python
#2
from datetime import datetime

# Create the two datetime objects
time1 = datetime(2026, 2, 15, 10, 0)
time2 = datetime(2026, 2, 15, 12, 30)

# Calculate the difference
diff = time2 - time1

# Extract hours and minutes from total seconds
total_seconds = int(diff.total_seconds())
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60

print(f"Difference: {hours} hours and {minutes} minutes")
```

```python
#3
from datetime import datetime

# Convert string to datetime object using strptime()
date_str = "20/02/2026"
dt_obj = datetime.strptime(date_str, "%d/%m/%Y")

# Format and display the result as YYYY-MM-DD using strftime()
formatted_date = dt_obj.strftime("%Y-%m-%d")
print(formatted_date)
```

```python
#4
import pandas as pd

dates = ["2026-02-01", "2026-02-05", "2026-02-10"]
pd_dates = pd.to_datetime(dates)

print(pd_dates)
```

```python
#5
import pandas as pd

dates = ["2026-02-01", "invalid", "2026-02-05", None]
pd_dates = pd.to_datetime(dates, errors="coerce")

# Identify missing/invalid values (NaT)
missing_values = pd_dates.isna()

print("Converted dates:")
print(pd_dates)
print("\nMissing value indicators:")
print(missing_values)
```

```python
#6
import pandas as pd

# Create the Series as specified for Questions 6-10
dates = pd.to_datetime(["2026-02-01", "2026-02-02", "2026-02-04", "2026-02-06", "2026-02-08"])
sales = pd.Series([100, 150, 120, 180, 200], index=dates, name="units_sold")

# Display index and its type
print("Index of sales series:")
print(sales.index)

print("\nIndex type:")
print(type(sales.index))
```

```python
# 7
# Retrieve the sales value on 4 February 2026 using .loc[]
sales_value = sales.loc["2026-02-04"]
print(f"Sales on 2026-02-04: {sales_value}")
```

```python
# 8
# Retrieve the first and last observations using .iloc[]
first_observation = sales.iloc[0]
last_observation = sales.iloc[-1]

print(f"First observation: {first_observation}")
print(f"Last observation: {last_observation}")
```

```python
# 9
# Select all sales observations between 2 February and 6 February 2026 inclusive
sales_slice = sales.loc["2026-02-02":"2026-02-06"]
display(sales_slice)
```

```python
# 10
# Calculate the total and average units sold during February 2026
total_sales = sales.sum()
average_sales = sales.mean()

print(f"Total units sold in February 2026: {total_sales}")
print(f"Average units sold per record: {average_sales}")
```

```python
# 11
# Generate 10 consecutive daily dates starting from 1 March 2026
daily_index = pd.date_range(start="2026-03-01", periods=10, freq="D")
print(daily_index)
```

```python
# 12
# Generate five daily dates ending on 15 March 2026
ending_index = pd.date_range(end="2026-03-15", periods=5, freq="D")
print(ending_index)
```

```python
# 13
# Generate six dates starting from 1 February 2026, with a two-day interval
two_day_index = pd.date_range(start="2026-02-01", periods=6, freq="2D")
print(two_day_index)
```

```python
# 14
# Generate hourly timestamps from 9:00 AM to 3:00 PM on 1 February 2026
hourly_index = pd.date_range(start="2026-02-01 09:00:00", end="2026-02-01 15:00:00", freq="h")
print(hourly_index)
```

```python
# 15
# Generate the last calendar day of each month from January to June 2026
end_of_month_index = pd.date_range(start="2026-01-01", end="2026-06-30", freq="ME")
print(end_of_month_index)
```
